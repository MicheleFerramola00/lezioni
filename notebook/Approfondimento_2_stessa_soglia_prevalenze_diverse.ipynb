### Passo 1 · Due gruppi di 1.000 pazienti

Stesso sistema: trova l'80% dei malati (recall) e riconosce il 90% dei sani (specificity).
Nel gruppo A la malattia colpisce il 20% dei pazienti, nel gruppo B il 5%.

**Atteso:** precision 67% in A e 30% in B.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def gruppo(nome, pazienti, frequenza, recall=0.80, specificity=0.90):
    malati = pazienti * frequenza
    sani = pazienti - malati
    trovati = recall * malati                    # veri positivi
    falsi_allarmi = (1 - specificity) * sani     # falsi positivi
    return {"gruppo": nome, "malati": malati, "malati trovati": trovati, "sani segnalati per errore": falsi_allarmi,
            "recall": trovati / malati, "specificity": (sani - falsi_allarmi) / sani,
            "precision": trovati / (trovati + falsi_allarmi)}

pd.DataFrame([gruppo("A, malattia al 20%", 1000, 0.20), gruppo("B, malattia al 5%", 1000, 0.05)]).round(2)
```

### Passo 2 · La precision segue la frequenza della malattia

Stesso sistema, stessa soglia; cambia solo quanto è frequente la malattia.

**Atteso:** sotto il 10% di malati, meno di un positivo su due è davvero malato.

```python
frequenze = np.linspace(0.01, 0.50, 50)
precision = [gruppo("", 1000, f)["precision"] for f in frequenze]
ax = plt.figure(figsize=(8, 3.8)).gca()
ax.plot(frequenze * 100, np.array(precision) * 100, color="#103F6D", lw=2.5)
for nome, f in (("A", 0.20), ("B", 0.05)):
    v = gruppo("", 1000, f)["precision"] * 100
    ax.plot(f * 100, v, "o", color="#D4A13A", ms=9); ax.text(f * 100 + 1, v - 6, f"{nome}: {v:.0f}%")
ax.set_xlabel("pazienti malati nel gruppo (%)"); ax.set_ylabel("precision (%)")
ax.set_title("Recall 80% e specificity 90% ovunque: la precision no")
plt.tight_layout(); plt.show()
```

---

### Cosa abbiamo visto

1. **Recall e specificity uguali non bastano:** la precision dipende anche da quanto è frequente la malattia.
2. **Per pareggiare la precision** bisognerebbe essere più severi nel gruppo B, e lì si troverebbero meno malati.
3. **La soglia si tara sulla popolazione in cui il sistema viene usato.**
